```python
# =====================================================
# STEP 33H — STORE PERFORMANCE RANKING
# =====================================================

store_ranking = (
    walmart_df
    .groupby("store")["weekly_sales"]
    .agg(["count", "mean", "sum"])
    .reset_index()
)

store_ranking.columns = [
    "store",
    "record_count",
    "average_sales",
    "total_sales"
]

# Rank stores by total sales
store_ranking["sales_rank"] = (
    store_ranking["total_sales"]
    .rank(method="dense", ascending=False)
    .astype(int)
)

# Overall average store sales
overall_store_average = store_ranking["total_sales"].mean()

# Performance classification
store_ranking["performance"] = np.where(
    store_ranking["total_sales"] > overall_store_average,
    "Above Average",
    "Below Average"
)

# Sort by rank
store_ranking = store_ranking.sort_values("sales_rank")

display(store_ranking.head(10))
```

```python
# =====================================================
# STEP 34 — STORE TYPE × STORE PERFORMANCE ANALYSIS
# =====================================================

store_type_performance = (
    walmart_df
    .groupby("type")
    .agg(
        store_count=("store", "nunique"),
        record_count=("weekly_sales", "count"),
        average_sales=("weekly_sales", "mean"),
        total_sales=("weekly_sales", "sum")
    )
    .reset_index()
)

store_type_performance.columns = [
    "store_type",
    "store_count",
    "record_count",
    "average_sales",
    "total_sales"
]

display(store_type_performance)
```

```python
# =====================================================
# STEP 35 — DEPARTMENT PERFORMANCE ANALYSIS
# =====================================================

department_performance = (
    walmart_df
    .groupby("dept")
    .agg(
        record_count=("weekly_sales", "count"),
        average_sales=("weekly_sales", "mean"),
        total_sales=("weekly_sales", "sum")
    )
    .reset_index()
)

# Rank departments by total sales
department_performance["sales_rank"] = (
    department_performance["total_sales"]
    .rank(method="dense", ascending=False)
    .astype(int)
)

# Sort by sales rank
department_performance = department_performance.sort_values(
    "sales_rank"
)

display(department_performance.head(10))
```

```python
# =====================================================
# STEP 39 — STORE SIZE vs SALES ANALYSIS
# =====================================================

store_size_sales = (
    walmart_df
    .groupby(["store", "type", "size"])
    .agg(
        record_count=("weekly_sales", "count"),
        average_sales=("weekly_sales", "mean"),
        total_sales=("weekly_sales", "sum")
    )
    .reset_index()
)

# Sort by store size
store_size_sales = store_size_sales.sort_values(
    "size",
    ascending=False
)

display(store_size_sales.head(15))
```

```python
# =====================================================
# STEP 40 — STORE SIZE CATEGORY ANALYSIS
# =====================================================

store_size_category = store_size_sales.copy()

store_size_category["size_category"] = pd.cut(
    store_size_category["size"],
    bins=[0, 100000, 180000, float("inf")],
    labels=["Small", "Medium", "Large"]
)

size_category_summary = (
    store_size_category
    .groupby("size_category", observed=True)
    .agg(
        store_count=("store", "nunique"),
        average_store_size=("size", "mean"),
        average_sales=("average_sales", "mean"),
        total_sales=("total_sales", "sum")
    )
    .reset_index()
)

display(size_category_summary)
```

```python
# =====================================================
# STEP 41 — STORE TYPE × SIZE CATEGORY ANALYSIS
# =====================================================

store_type_size = store_size_category.copy()

type_size_summary = (
    store_type_size
    .groupby(["type", "size_category"], observed=True)
    .agg(
        store_count=("store", "nunique"),
        average_store_size=("size", "mean"),
        average_sales=("average_sales", "mean"),
        total_sales=("total_sales", "sum")
    )
    .reset_index()
)

display(type_size_summary)
```

```python
# =====================================================
# STEP 42 — TEMPERATURE vs WEEKLY SALES
# =====================================================

temperature_sales = (
    walmart_df[
        ["temperature", "weekly_sales"]
    ]
    .copy()
)

temperature_sales["temperature_range"] = pd.cut(
    temperature_sales["temperature"],
    bins=[-float("inf"), 32, 50, 70, 90, float("inf")],
    labels=[
        "Very Cold",
        "Cold",
        "Moderate",
        "Warm",
        "Very Warm"
    ]
)

temperature_summary = (
    temperature_sales
    .groupby("temperature_range", observed=True)
    .agg(
        record_count=("weekly_sales", "count"),
        average_sales=("weekly_sales", "mean"),
        total_sales=("weekly_sales", "sum")
    )
    .reset_index()
)

display(temperature_summary)
```

```python
# =====================================================
# STEP 43 — FUEL PRICE vs WEEKLY SALES
# =====================================================

fuel_price_sales = (
    walmart_df[
        ["fuel_price", "weekly_sales"]
    ]
    .copy()
)

fuel_price_sales["fuel_price_range"] = pd.cut(
    fuel_price_sales["fuel_price"],
    bins=[0, 2.75, 3.00, 3.50, 4.00, float("inf")],
    labels=[
        "Low",
        "Moderate-Low",
        "Moderate-High",
        "High",
        "Very High"
    ]
)

fuel_price_summary = (
    fuel_price_sales
    .groupby("fuel_price_range", observed=True)
    .agg(
        record_count=("weekly_sales", "count"),
        average_sales=("weekly_sales", "mean"),
        total_sales=("weekly_sales", "sum")
    )
    .reset_index()
)

display(fuel_price_summary)
```

```python
# =====================================================
# STEP 44 — MARKDOWN IMPACT ANALYSIS
# =====================================================

markdown_columns = [
    "markdown1",
    "markdown2",
    "markdown3",
    "markdown4",
    "markdown5"
]

# Calculate total markdown
walmart_df["total_markdown"] = walmart_df[markdown_columns].sum(axis=1)

# Create markdown status
walmart_df["markdown_status"] = np.where(
    walmart_df["total_markdown"] > 0,
    "Markdown Applied",
    "No Markdown"
)

# Compare sales performance
markdown_summary = (
    walmart_df
    .groupby("markdown_status")["weekly_sales"]
    .agg(
        record_count="count",
        average_sales="mean",
        total_sales="sum"
    )
    .reset_index()
)

display(markdown_summary)
```

```python
# =====================================================
# STEP 45 — NEGATIVE & ZERO SALES ANALYSIS
# =====================================================

walmart_df["sales_category"] = np.select(
    [
        walmart_df["weekly_sales"] < 0,
        walmart_df["weekly_sales"] == 0,
        walmart_df["weekly_sales"] > 0
    ],
    [
        "Negative Sales",
        "Zero Sales",
        "Positive Sales"
    ],
    default="Unknown"
)

sales_quality_summary = (
    walmart_df
    .groupby("sales_category")["weekly_sales"]
    .agg(
        record_count="count",
        average_sales="mean",
        total_sales="sum"
    )
    .reset_index()
)

display(sales_quality_summary)
```

```python
# =====================================================
# STEP 46 — WEEKLY SALES OUTLIER ANALYSIS
# =====================================================

Q1 = walmart_df["weekly_sales"].quantile(0.25)
Q3 = walmart_df["weekly_sales"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

outlier_mask = (
    (walmart_df["weekly_sales"] < lower_bound) |
    (walmart_df["weekly_sales"] > upper_bound)
)

outlier_count = outlier_mask.sum()
normal_count = (~outlier_mask).sum()

print("Q1:", round(Q1, 2))
print("Q3:", round(Q3, 2))
print("IQR:", round(IQR, 2))

print("\nLower Bound:", round(lower_bound, 2))
print("Upper Bound:", round(upper_bound, 2))

print("\nOutlier Records:", outlier_count)
print("Normal Records:", normal_count)
```

```python
# =====================================================
# STEP 47 — OUTLIER SALES SUMMARY
# =====================================================

outlier_df = walmart_df[outlier_mask].copy()

outlier_summary = pd.DataFrame({
    "metric": [
        "Outlier Records",
        "Average Outlier Sales",
        "Minimum Outlier Sales",
        "Maximum Outlier Sales",
        "Total Outlier Sales"
    ],
    "value": [
        len(outlier_df),
        outlier_df["weekly_sales"].mean(),
        outlier_df["weekly_sales"].min(),
        outlier_df["weekly_sales"].max(),
        outlier_df["weekly_sales"].sum()
    ]
})

display(outlier_summary)
```

```python
# =====================================================
# STEP 48 — CORRELATION ANALYSIS
# =====================================================

correlation_columns = [
    "weekly_sales",
    "temperature",
    "fuel_price",
    "markdown1",
    "markdown2",
    "markdown3",
    "markdown4",
    "markdown5",
    "size"
]

correlation_matrix = (
    walmart_df[correlation_columns]
    .corr()
    .round(3)
)

display(correlation_matrix)
```

```python
# =====================================================
# STEP 49 — FINAL PYTHON KPI SUMMARY
# =====================================================

total_sales = walmart_df["weekly_sales"].sum()
average_sales = walmart_df["weekly_sales"].mean()

total_records = len(walmart_df)
total_stores = walmart_df["store"].nunique()
total_departments = walmart_df["dept"].nunique()

minimum_sales = walmart_df["weekly_sales"].min()
maximum_sales = walmart_df["weekly_sales"].max()

negative_sales_count = (walmart_df["weekly_sales"] < 0).sum()
zero_sales_count = (walmart_df["weekly_sales"] == 0).sum()

holiday_average = (
    walmart_df.loc[
        walmart_df["isholiday"] == True,
        "weekly_sales"
    ].mean()
)

non_holiday_average = (
    walmart_df.loc[
        walmart_df["isholiday"] == False,
        "weekly_sales"
    ].mean()
)

kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Records",
        "Total Stores",
        "Total Departments",
        "Total Weekly Sales",
        "Average Weekly Sales",
        "Minimum Weekly Sales",
        "Maximum Weekly Sales",
        "Negative Sales Records",
        "Zero Sales Records",
        "Holiday Average Sales",
        "Non-Holiday Average Sales"
    ],
    "Value": [
        total_records,
        total_stores,
        total_departments,
        total_sales,
        average_sales,
        minimum_sales,
        maximum_sales,
        negative_sales_count,
        zero_sales_count,
        holiday_average,
        non_holiday_average
    ]
})

display(kpi_summary)
```

```python
# =====================================================
# STEP 52 — CREATE FINAL PYTHON ANALYSIS DATASET
# =====================================================

final_store_analysis = (
    walmart_df
    .groupby(["store", "type", "size"])["weekly_sales"]
    .agg(
        record_count="count",
        average_sales="mean",
        total_sales="sum"
    )
    .reset_index()
)

final_store_analysis["sales_rank"] = (
    final_store_analysis["total_sales"]
    .rank(method="dense", ascending=False)
    .astype(int)
)

final_store_analysis = final_store_analysis.sort_values(
    "sales_rank"
)

display(final_store_analysis.head(15))
```
